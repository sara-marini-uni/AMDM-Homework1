```python
# Exercise 12: The Minion Game
def minion_game(string):
    vowels = 'AEIOU'
    kevin_score = 0
    stuart_score = 0
    length = len(string)
    
    for i in range(length):
        if string[i] in vowels:
            kevin_score += length - i
        else:
            stuart_score += length - i
            
    if kevin_score > stuart_score:
        print("Kevin", kevin_score)
    elif stuart_score > kevin_score:
        print("Stuart", stuart_score)
    else:
        print("Draw")

if __name__ == '__main__':
    s = input()
    minion_game(s)
```

```python
# Exercise 8: Piling up!
t = int(input())
for _ in range(t):
    n = int(input())
    cubes = list(map(int, input().split()))
    left = 0
    right = n - 1
    last_cube = float('inf')
    possible = True
    
    while left <= right:
        if cubes[left] >= cubes[right]:
            if cubes[left] <= last_cube:
                last_cube = cubes[left]
                left += 1
            else:
                possible = False
                break
        else:
            if cubes[right] <= last_cube:
                last_cube = cubes[right]
                right -= 1
            else:
                possible = False
                break
    
    print("Yes" if possible else "No")
```

```python
# Exercise 2: Time Delta
import math
import os
import random
import re
import sys

def time_delta(t1, t2):
    from datetime import datetime, timedelta
    format_str = '%a %d %b %Y %H:%M:%S %z'
    dt1 = datetime.strptime(t1, format_str)
    dt2 = datetime.strptime(t2, format_str)
    delta = abs(dt1 - dt2)
    return str(int(delta.total_seconds()))

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    t = int(input())

    for t_itr in range(t):
        t1 = input()

        t2 = input()

        delta = time_delta(t1, t2)

        fptr.write(delta + '\n')

    fptr.close()
```

```python
# Exercise 1: XML 1 - Find the Score
import sys
import xml.etree.ElementTree as etree
def get_attr_number(node):
    count = len(node.attrib)
    for child in node:
        count += get_attr_number(child)
    return count
if __name__ == '__main__':
    sys.stdin = open('input.xml', 'r')
    xml = sys.stdin.read()
    tree = etree.ElementTree(etree.fromstring(xml))
    root = tree.getroot()
    print(get_attr_number(root))
```

```python
# Exercise 2: XML 2 - Find the Maximum Depth
import xml.etree.ElementTree as etree
maxdepth = 0
def depth(elem, level):
    global maxdepth
    maxdepth = max(maxdepth, level + 1)
    for child in elem:
        depth(child, level + 1)
if __name__ == '__main__':
    n = int(input())
    xml = ""
    for i in range(n):
        xml =  xml + input() + "\n"
    tree = etree.ElementTree(etree.fromstring(xml))
    depth(tree.getroot(), -1)
    print(maxdepth)
```

```python
# Exercise 2: Decorators 2 - Name Directory
import operator
def person_lister(f):
    def inner(people):
        people.sort(key=lambda x: int(x[2]))
        return map(f, people)
    return inner
@person_lister
def name_format(person):
    return ("Mr. " if person[3] == "M" else "Ms. ") + person[0] + " " + person[1]

if __name__ == '__main__':
    people = [input().split() for i in range(int(input()))]
    print(*name_format(people), sep='\n')
```
